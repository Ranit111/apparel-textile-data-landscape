# Apparel & Textiles — Real Public Data EDA
This notebook uses the included official EEA Circularity Metrics Lab extract.

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

path = Path('../data/raw/eea_textile_indicators.csv')
df = pd.read_csv(path)
df
```

```python
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicates:', df.duplicated().sum())
```

```python
pct = df[df['unit'] == '%'].sort_values('value', ascending=False)
pct[['indicator','year','value','unit']]
```

```python
plt.figure(figsize=(9,5))
plt.barh(pct['indicator'], pct['value'])
plt.xlabel('Percent (%)')
plt.title('Selected EU Textile Circularity Indicators')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Interpretation
The low household textile-waste capture rate is a useful operational KPI. The treatment share and raw-material-use indicators provide additional circularity and resource-risk context. These indicators should be joined with trade/production datasets only after scope and methodology alignment.
